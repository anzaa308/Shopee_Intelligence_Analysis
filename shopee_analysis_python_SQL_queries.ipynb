```python
import numpy as np

# 1. Clean 'total_sold' column (convert strings like '8.1k' to numeric floats)
def clean_sold(val):
    if pd.isna(val):
        return 0.0
    val_str = str(val).lower().strip()
    if 'k' in val_str:
        return float(val_str.replace('k', '')) * 1000
    try:
        return float(val_str)
    except:
        return 0.0

df['total_sold_clean'] = df['total_sold'].apply(clean_sold)

# 2. Create a Discount Percentage column (comparing original vs actual price)
# Fill missing original prices with actual prices to avoid NaN errors
df['price_ori_filled'] = df['price_ori'].fillna(df['price_actual'])
df['discount_pct'] = ((df['price_ori_filled'] - df['price_actual']) / df['price_ori_filled']) * 100

# 3. Quick check of our new clean columns
print(df[['title', 'price_actual', 'total_sold_clean', 'discount_pct']].head())
```

```python
import sqlite3

```

```python
conn = sqlite3.connect(':memory:')
```

```python
df.to_sql('shopee_products', conn, index=False, if_exists='replace')
```

```python
query = """
SELECT
    item_category_detail,
    COUNT(*) as total_listings,
    ROUND(AVG(price_actual), 2) as avg_price,
    ROUND(SUM(total_sold_clean), 0) as cumulative_units_sold
FROM shopee_products
GROUP BY item_category_detail
ORDER BY cumulative_units_sold DESC
LIMIT 10;
"""

```

```python
top_categories_sql = pd.read_sql(query, conn)
print(top_categories_sql.to_string())
```
